# 부산항 컨테이너 성장에서 환적의 기여 구조 파악

```python
df=pd.read_csv('cct.csv', encoding='cp949')

df.head()
```

```python
# 연도별 환적점유율
trans_share = df[['연도', '환적점유율']]

trans_share
```

```python
print(df.columns)
df.columns = df.columns.str.strip()
```

```python
# 환적 기여율

#전년 대비 증가량
df['전체증가량'] = df['컨물동량'].diff()
df['환적증가량'] = df['환적소계'].diff()

# 환적 기여율
df['환적기여율(%)'] = (
    df['환적증가량'] / df['전체증가량'] * 100
).round(2)

df[['연도', '전체증가량', '환적증가량', '환적기여율(%)']]
```

전체 물동량이 증가한 연도 대부분에서 환적기여율이 높은 수준을 보임
일부 연도에는 100%를 넘어 수출입 물동량 감소를 환적 증가가 상쇄한 것으로 나타남
따라서 부산항의 물동량 성장은 환적 물동량 증가에 크게 의존하는 구조라고 볼 수 있다.

```python
#환적과 수출입 중 누가 더 많이 성장했는지

#수출입증감률
df['수출입증감률'] = (
    df['수출입소계'].pct_change() * 100
).round(2)

df[['연도', '수출입증감률', '환적증감율']]
```

2014~2025년 부산항은 수출입보다 환적 물동량의 성장률이 전반적으로 높게 나타남
수출입이 감소한 시기에도 환적이 증가하는 모습을 보여 부산항의 물동량 성장이 환적을 중심으로 이루어지는 경향을 확인할 수 있음
