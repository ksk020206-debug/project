# 장기성장 
1997년부터 꾸준히 성장, 2009년에 하락

```python
df = pd.read_csv('Forecast.csv', encoding='cp949')
df.head()

monthly = (df.groupby('연월')['물동량'].sum().reset_index())

monthly
```

```python
monthly = (
df.groupby(['연도', '월'])['물동량'] .sum() .reset_index()
)

monthly_avg = (monthly.groupby('월')['물동량'].mean().reset_index()
)

monthly_avg['증감률(%)'] = monthly_avg['물동량'].pct_change() * 100

monthly_avg
```

```python
avg = monthly_avg['물동량'].mean()

monthly_avg['평균차이'] = monthly_avg['물동량'] - avg

monthly_avg
```

```python
monthly_avg['차이율(%)'] = (monthly_avg['평균차이'] / avg * 100).round(2)
monthly_avg
```
